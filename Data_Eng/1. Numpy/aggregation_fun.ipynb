```python
import numpy as np
arr = np.array([10,20,30,40,50])
print(np.sum(arr))
print(np.mean(arr))
print(np.median(arr))
print(np.min(arr))
print(np.max(arr))
print(np.std(arr))
print(np.var(arr))
print(np.argmin(arr))
print(np.argmax(arr))
```

```python
#Aggregation on 2d

arr= np.array([[1,2,3],[4,5,6]]
              )

print(np.sum(arr))
print(np.sum(arr,axis=0))
print(np.sum(arr,axis=1))
```

```python
#Boolean Indexing

scores = np.random.randint(1,100,size=(10))
print(scores)
print(scores.shape)
mask = scores >60
print(mask)
print(scores[mask])
```

```python
mask2 = (scores>70) & (scores<85)
print(mask2)
```
